```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv("Daily_Cal_Req.csv")
print(df.head())
```

```python
df['Age_Group']=pd.cut(df['Age'],bins=[0,20,30,40,50,60],
                       labels=["<=20","21-30","31-40","41-50","51-60"])
for col in ["Diet_Type","Activity_Level","Age_Group","Gender"]:
    if col in df.columns:
        print(df[col].value_counts())
```

```python
if set(["Goal","Daily_Calorie_Requirement"]).issubset(df.columns):
    print(df.groupby("Goal")["Daily_Calorie_Requirement"].mean())
```

```python
if set(["Age","Daily_Calorie_Requirement"]).issubset(df.columns):
    sns.scatterplot(data=df,x="Age",y="Daily_Calorie_Requirement")
plt.show()
```

```python
if set(["Activity_Level","Water_Intake_L"]).issubset(df.columns):
    sns.barplot(data=df,x="Activity_Level",y="Water_Intake_L")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.barplot(x='Activity_Level',y='Daily_Calorie_Requirement',hue='Gender',data=df)
plt.title('Calories by Activity level and Gender')
plt.show()
```

```python
data=df.copy()
le=LabelEncoder()
for col in data.select_dtypes(include='object').columns:
    data[col]=le.fit_transform(data[col].astype(str))
plt.figure(figsize=(10,8))
sns.heatmap(data.corr(), annot=True,cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
